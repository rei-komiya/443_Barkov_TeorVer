### Задача 1.8
> В конкурсе по 5 номинациям участвуют 10 кинофильмов. Сколько существует вариантов распределения призов, если по каждой номинации установлены: а) различные призы; б) одинаковые призы?

```python
movies = range(10)
```

```python
all_variants_a = list(itertools.product(movies, repeat=5))
print(f'а) {len(all_variants_a)}')
```

```python
all_variants_b = list(itertools.combinations_with_replacement(movies, 5))
print(f'б) {len(all_variants_b)}')
```

### Задача 1.10
> Буквы Т, Е, И, Я, Р, О написаны на отдельных карточках. Ребенок берет карточки в случайном порядке и прикладывает одну к друтой: а) 3 карточки; б) все 6 карточек. Какова вероятность того, что получится слово: а) «ТОР»; б) «ТЕОРИЯ»?

### Задача 1.16
> За круглым столом рассаживаются 5 мужчин и 5 женщин. Найти вероятность того, что: а) никакие два лица одного пола не сядут рядом; б) супруги сядут рядом, если эти мужчины и женщины образуют 5 супружеских пар.

```python
men = [0, 2, 4, 6, 8]
women = [1, 3, 5, 7, 9]

total = 0
count_a = 0
count_b = 0

for p in list(itertools.permutations(men + women)):
    total += 1
    
    for i in range(9):
        if p[i] % 2 == p[i + 1] % 2:
            break
    else:
        count_a += 1

    pairs = 0
    for i in range(9):
        if p[i] // 2 == p[i + 1] // 2:    # Считаем, что пары - (0, 1), (2, 3), (4, 5) и т.д.
            pairs += 1
    if pairs == 5:
        count_b += 1

print(f'а) {round(count_a / total, 4)}')
print(f'б) {round(count_b / total, 4)}')
        
```

### Задача 1.17
> В купейный вагон (9 купе по 4 места) семи пассажирам продано 7 билетов. Найти вероятности того, что пассажиры попали: а) в два купе; б) в семь купе; в) в три купе.

```python
import math

wagon = range(5) # варианты занятых мест в вагоне

total = 0
count_a = 0
count_b = 0
count_c = 0

for p in list(itertools.product(wagon, repeat=9)):
    if sum(p) == 7:
        total_occupied = sum(1 for i in p if i > 0)
    
        ways = 1
        for i in p:
            ways *= math.comb(4, i) # находим все способы распределения мест среди всех пассажиров

        total += ways
        if total_occupied == 2:
            count_a += ways
        elif total_occupied == 7:
            count_b += ways
        elif total_occupied == 3:
            count_c += ways

print(f'а) {round(count_a / total, 7)}')
print(f'б) {round(count_b / total, 4)}')
print(f'в) {round(count_c / total, 4)}')
```
